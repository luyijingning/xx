git statuus//查看仓库状态  
git init//初始化仓库  
git add name//添加文件 git add .添加所有文件  
git commit -m "commit text"//提交添加的文件  
git log //打印日志  
git push origin master//推送文件  
git pull origin master//拉取仓库  
git remote  add origin url//连接远程仓库  
git branch //查看分支情况 git branch a 创建a分支  
git checkout a //更换到a分支 git checkout -b c //创建并切换到c分支    
git merge a//在master分支下合并a分支  
git branch -d a//删除a分支，-D 强制删除  
git tag //为当前标签添加标签 git tag V1.0
