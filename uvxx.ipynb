uv学习命令：
uv venv --python 3.12  
uv venv #在当前文件夹下创建虚拟环境  
source .venv/bin/activate#激活虚拟环境  
2.包管理  
uv pip install requests#安装最新版本  
uv pip install requests==2.31.0#安装特定版本  
uv pip install -r requirements.txt  
uv pip install --upgrade requests  
uv pip uninstall requests  
uv pip freeze > requirements.txt    
uv pip list#查看已安装的包  
3.项目管理 以pyproject.toml为中心  
uv init my_project#初始化项目  
以下为项目基本结构：  
my_project/  
├── pyproject.toml    # 项目配置和依赖声明
├── .python-version   # 固定 Python 版本
├── README.md  
└── main.py  

# 添加生产依赖
uv add requests

# 添加指定版本的依赖
uv add "requests>=2.31.0"

# 添加开发依赖（只在开发环境使用，如测试框架）
uv add --dev pytest ruff

# 移除依赖
uv remove requests

uv sync#安装或更新pyproject.toml后用于安装所有依赖
# 直接运行 Python 脚本
uv run main.py

# 运行项目中的测试
uv run pytest

# 运行任意命令（在虚拟环境的上下文中执行）
uv run python -c "import requests; print(requests.__version__)"
